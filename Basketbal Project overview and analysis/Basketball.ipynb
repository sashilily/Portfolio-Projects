```python
import pandas as pd
import glob
import os
```

```python
data = {}

# 1. Path to your folder (change 'my_folder_name' to yours)
folder_path = r"C:\Users\user\Documents\DataScience\basketball-project-workspace\march-machine-learning-mania-2026"

# 2. Get a list of all CSV files in that folder
files = glob.glob(os.path.join(folder_path, "*.csv"))

data = {}
for file in files:
    # 3. Use os.path.basename to get the name without the folder path
    name = os.path.basename(file).replace(".csv", "")
    data[name] = pd.read_csv(file)
len(data)
```

```python
# See everything in the folder to check extensions
print(os.listdir(folder_path)) 
```

```python
for name, df in data.items():
    print(f"--- Dataset: {name} ---")
    print(f"Columns: {df.columns}")
    df.info()
    print("\n")
```

```python
total_cells = sum(df.size for df in data.values())
total_cells
```

```python
print(f"Total Files: {len(data)}")
print(f"Total Records: {sum(len(df) for df in data.values()):,}")
print(f"Total Columns: {sum(df.shape[1] for df in data.values()):,}")
```
