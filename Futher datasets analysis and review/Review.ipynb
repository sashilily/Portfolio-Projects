```python
import numpy as np
monthly_sales= np.array([[ 4134, 23925,  8657],
       [ 4116, 23875,  9142],
       [ 4673, 27197, 10645],
       [ 4580, 25637, 10456],
       [ 5109, 27995, 11299],
       [ 5011, 27419, 10625],
       [ 5245, 27305, 10630],
       [ 5270, 27760, 11550],
       [ 4680, 24988,  9762],
       [ 4913, 25802, 10456],
       [ 5312, 25405, 13401],
       [ 6630, 27797, 18403]])
```

```python
# Create the 1D array avg_monthly_sales
avg_monthly_sales = monthly_sales.mean(axis=1)
print(avg_monthly_sales)

# Plot avg_monthly_sales by month
plt.plot(np.arange(1,13), avg_monthly_sales, label="Average sales across industries")

# Plot department store sales by month
plt.plot(np.arange(1,13), monthly_sales[:,2], label="Department store sales")
plt.legend()
plt.show()
```

```python
# Find cumulative monthly sales for each industry
cumulative_monthly_industry_sales = monthly_sales.cumsum(axis=0)
print(cumulative_monthly_industry_sales)

# Plot each industry's cumulative sales by month as separate lines
plt.plot(np.arange(1, 13), cumulative_monthly_industry_sales[:,0], label="Liquor Stores")
plt.plot(np.arange(1, 13), cumulative_monthly_industry_sales[:,1], label="Restaurants")
plt.plot(np.arange(1, 13), cumulative_monthly_industry_sales[:,2], label="Department stores")
plt.legend()
plt.show()
```

```python
monthly_industry_multipliers = np.array([
       [0.98, 1.02, 1.  ],
       [1.00, 1.01, 0.97],
       [1.06, 1.03, 0.98],
       [1.08, 1.01, 0.98],
       [1.08, 0.98, 0.98],
       [1.1 , 0.99, 0.99],
       [1.12, 1.01, 1.  ],
       [1.1 , 1.02, 1.  ],
       [1.11, 1.01, 1.01],
       [1.08, 0.99, 0.97],
       [1.09, 1.  , 1.02],
       [1.13, 1.03, 1.02]])
```

```python
# Create an array of monthly projected sales for all industries
projected_monthly_sales = monthly_sales * monthly_industry_multipliers
print(projected_monthly_sales)

# Graph current liquor store sales and projected liquor store sales by month
plt.plot(np.arange(1,13), monthly_sales[:,0], label="Current liquor store sales")
plt.plot(np.arange(1,13), projected_monthly_sales[:,0], label="Projected liquor store sales")
plt.legend()
plt.show()
```

```python
def average(values):
    """
    Find the mean in a sequence of values and round to two decimal places.
    Args:
        values (list): A list of numeric values.
        
    Returns:
        rounded_average (float): The mean of values, rounded to two decimal places.
    """
    aver_val = sum(values)/ len(values)
    round_aver = round(aver_val, 2)
    return round_aver
```

```python
help(average)
```

```python
average([15, 29, 4, 13, 11, 8])
```

```python
def average(*args):
    aver_val = sum(args)/ len(args)
    round_aver = round(aver_val, 2)
    return round_aver
```

```python
average(15, 29, 4, 13, 11, 8)
```

```python
average(*[15, 29], *[4, 13], *[11, 8])
```

```python
def average(**kwargs):
    aver_val = sum(kwargs.values())/ len(kwargs.values())
    round_aver = round(aver_val, 2)
    return round_aver
```

```python
average(a=15, b=29, c=4, d=13, e=11, f=8)
```

```python
average(**{'a':15, 'b':29, 'c':4, 'd':13, 'e':11, 'f':8})
```

### Lambda Functions

##### Simple functions

```python
def average(values):
    aver_val = sum(values)/ len(values)
    round_aver = round(aver_val, 2)
    return round_aver
```

```python
# Lambda average function
print(lambda x: sum(x) / len(x))
```

```python
# Custome average funcion
def average(x):
    return sum(x) / len(x)

print(average)
```

```python
# Print the average
print((lambda x: sum(x) / len(x)) ([3, 6, 9]))
```

```python
# Storing and calling lambda function
# Store as a variable
average = lambda x: sum(x) / len(x)
print(average([3, 6, 9]))
```

```python
# Multiple parameters

# with two args
power = lambda x, y: x**y

#Raise 2 to the power of 3
print(power(2,3))
```

```python
# with iterables

names = ['john', 'sally', 'leah']
# Apply a lambda fxn inside map()
capitalize = map(lambda x: x.capitalize(), names)
print(capitalize)
```

```python
# Convert to a list
print(list(capitalize))
```
