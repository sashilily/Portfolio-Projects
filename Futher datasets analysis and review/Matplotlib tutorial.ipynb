```python
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.931]
```

```python
plt.plot(yield_apples);
```

```python
import numpy as np
years = np.arange(2010,2015)
```

```python
plt.plot(years, yield_apples);
plt.xticks(np.arange(int(min(years)), int(max(years)) + 1, 1));
```

```python
plt.plot(years, yield_apples);
plt.xticks(np.arange(int(min(years)), int(max(years)) + 1, 1));
plt.xlabel('Year');
plt.ylabel('Yields of apple');
```

```python
years = np.arange(2010, 2021)
years
```

```python
apples = np.sort(np.random.rand(len(years)))
apples
```

```python
oranges = np.sort(np.random.rand(len(years)))
oranges
```

```python
plt.plot(years, apples)
plt.plot(years, oranges)


plt.xticks(np.arange(int(min(years)), int(max(years)) + 1, 1))
plt.xlabel('Year')
plt.ylabel('Yields (ton per hectare)');
```

```python





plt.plot(years, apples, label='Apples')
plt.plot(years, oranges, label= 'Oranges')


plt.xticks(np.arange(int(min(years)), int(max(years)) + 1, 1))
plt.xlabel('Year')
plt.ylabel('Yields (ton per hectare)')

plt.title('Crop yields')
plt.legend()
```

- lw: linewidth
- ms: markersize
- mec: markeredgecolor
- mew: markeredgewidth
- mfc: markerfacecolor
- alpha: opacity of the plot
- ls: linestyle

```python
plt.plot(years, apples, label='Apples', marker='o',c='g', ls='-',lw=2, ms=8, mew=2,mec='navy')
plt.plot(years, oranges, label= 'Oranges', marker='x', c='r',ls='--',lw=3, ms=10, alpha=.5)


plt.xticks(np.arange(int(min(years)), int(max(years)) + 1, 1))
plt.xlabel('Year')
plt.ylabel('Yields (ton per hectare)')



plt.title('Crop yields')
plt.legend()
```
