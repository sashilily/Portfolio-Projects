```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
tips.head(10)
```

Our tips dataset contains two columns 'total_bill' and 'tip' which was collected. We'd like to determine the relationship of these columns to answer the question "Does total_bill"and "tip" have any relationship? Are they dependent?

A scatter plot displays values for two variables as a sequence of coordinates plane.

```python
sns.set_theme(style='darkgrid')
sns.scatterplot(data=tips, x='total_bill', y='tip')

plt.title('Total Bill vs Tip')
plt.savefig('Total Bill vs Tip.png', dpi=300)

plt.show()
```

```python
# Calculate the correlation coefficient to identify the 
# linear relationship  of the two variables
r = tips['total_bill'].corr(tips['tip'])
round(r, 2)
```
