# Foundations of Probabilities in Python
### Flip a coin in Python

```python
from scipy.stats import bernoulli
bernoulli.rvs(p=0.5, size=1)
```

```python
bernoulli.rvs(p=0.5, size=10)
```

```python
sum(bernoulli.rvs(p=0.5, size=10))
```

```python
# Flipping Multiple coins
from scipy.stats import binom
binom.rvs(n=10, p=0.5, size=1)
```

```python
# Many draws
binom.rvs(n=10, p=0.5, size=10)
```

```python
# Biased coin draws
sum(binom.rvs(n=10, p=0.3, size=10))
```

```python
# Random generator seed
# use random_state parameter of the rvs() function
binom.rvs(n=10, p=0.5, size=1, random_state=42)
```
